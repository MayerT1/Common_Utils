```python
import rasterio
from rasterio.windows import Window
import numpy as np

# =====================================================
# USER INPUTS
# =====================================================
raster_path = "/content/RH_MLBS_2024.tif"  # Change this

# Center pixel (column, row)
center_col = 671
center_row = 875

# Window size (pixels)
window_width = 50
window_height = 50
# =====================================================

# Calculate upper-left corner of the window
col_off = center_col - window_width // 2
row_off = center_row - window_height // 2

window = Window(col_off, row_off, window_width, window_height)

with rasterio.open(raster_path) as src:
    # Read all bands in the window
    data = src.read(window=window)

    print("=" * 60)
    print(f"Raster: {raster_path}")
    print(f"Bands: {src.count}")
    print(f"Window size: {window_width} x {window_height}")
    print(f"Center pixel: (col={center_col}, row={center_row})")
    print("=" * 60)

    # Print each band as a matrix
    for b in range(src.count):
        print(f"\nBand {b+1}")
        print(data[b])

    print("\n" + "=" * 60)
    print("Per-pixel values (all bands)")
    print("=" * 60)

    bands, rows, cols = data.shape

    for r in range(rows):
        for c in range(cols):
            raster_row = row_off + r
            raster_col = col_off + c
            values = data[:, r, c]

            print(
                f"Raster Pixel (row={raster_row:5d}, col={raster_col:5d}) "
                f"Window({r:2d},{c:2d}) -> {values.tolist()}"
            )
```
