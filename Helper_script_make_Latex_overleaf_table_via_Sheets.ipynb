```python
import pandas as pd

# Replace this path with the path to your CSV in your mounted drive
csv_path = "/content/drive/MyDrive/PhD_Main_Folder/Geo_Data/Sewanee_DL_Project/GEDI_target/filtered_csvs/GEDI_filtered_merged_summary.csv"

# Read CSV
df = pd.read_csv(csv_path)
df.head()
```

```python
def df_to_latex(df, caption="My Table", label="tab:mytable"):
    """
    Convert a pandas DataFrame to a LaTeX table string ready for Overleaf.

    Args:
        df (pd.DataFrame): DataFrame to convert
        caption (str): Table caption
        label (str): Table label for referencing in LaTeX

    Returns:
        str: LaTeX table as a string
    """
    header = " & ".join(df.columns) + " \\\\ \\hline\n"
    rows = ""
    for i, row in df.iterrows():
        row_str = " & ".join([str(x) for x in row.values]) + " \\\\\n"
        rows += row_str

    latex_table = f"""
\\begin{{table}}[ht]
\\centering
\\caption{{{caption}}}
\\label{{{label}}}
\\begin{{tabular}}{{{' | '.join(['l']*len(df.columns))}}}
\\hline
{header}{rows}\\hline
\\end{{tabular}}
\\end{{table}}
"""
    return latex_table
```

```python
latex_code = df_to_latex(df, caption="GEDI Filtered Summary", label="tab:GEDI_summary")
print(latex_code)
```
